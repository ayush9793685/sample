```python
import matplotlib.pyplot as plt

def price_distribution_for_hsn_line(df, hsn_code):
    # Filter rows for the chosen HSN
    subset = df.loc[df['HsnCode'] == hsn_code, 'SpecialPrice']

    if subset.empty:
        print(f"No products found for HSN {hsn_code}")
        return

    # Count frequency of each selling price
    price_counts = subset.value_counts().sort_index()

    # Plot line graph
    plt.figure(figsize=(12,6))
    plt.plot(price_counts.index, price_counts.values, marker='o')
    plt.title(f"Price Frequency Distribution (Line Graph) for HSN Code {hsn_code}")
    plt.xlabel("Selling Price")
    plt.ylabel("Number of Products at this Price")
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.show()

    # Print the most frequent selling price
    most_frequent_price = price_counts.idxmax()
    count = price_counts.max()
    print(f" The most common selling price for HSN {hsn_code} is {most_frequent_price} (appears {count} times).")
```

```python
import matplotlib.pyplot as plt
import numpy as np

for hsn in top_hsns_list:
    print(f"\nAnalyzing HSN Code: {hsn}")
    hsn_subset = filtered_merged_df[filtered_merged_df['hsncode'] == hsn].sort_values(by='total order', ascending=False)

    if hsn_subset.empty:
        print(f"No products found for HSN {hsn} in the filtered data.")
        continue

    max_orders = hsn_subset['total order'].max()
    top_sellers = hsn_subset[hsn_subset['total order'] == max_orders]

    # Select a few lower sellers for comparison (e.g., top 5 lower sellers)
    lower_sellers = hsn_subset[hsn_subset['total order'] < max_orders].head(5)

    print(f"\nTop Selling Product(s) for HSN {hsn} (Total Orders: {max_orders}):")
    display(top_sellers[['productsku', 'productname', 'specialprice', 'categoryname', 'sellerstate', 'total order']])

    if not lower_sellers.empty:
        print(f"\nLower Selling Product(s) for HSN {hsn}:")
        display(lower_sellers[['productsku', 'productname', 'specialprice', 'categoryname', 'sellerstate', 'total order']])

        # Combine top sellers and lower sellers for plotting
        products_to_plot = pd.concat([top_sellers, lower_sellers])

        # Create a bar plot for Total Orders
        plt.figure(figsize=(12, 6))
        plt.bar(products_to_plot['productsku'], products_to_plot['total order'], color='skyblue')
        plt.ylabel('Total Orders')
        plt.title(f'Total Orders for Top and Lower Selling Products (HSN {hsn})')
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()

        # Create a bar plot for Special Price
        plt.figure(figsize=(12, 6))
        plt.bar(products_to_plot['productsku'], products_to_plot['specialprice'], color='lightcoral')
        plt.ylabel('Special Price')
        plt.title(f'Selling Price for Top and Lower Selling Products (HSN {hsn})')
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()

    else:
        print(f"\nNo lower selling products found for HSN {hsn} to compare.")

    print("-" * 50)
```
